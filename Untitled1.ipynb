```python
# ============================================================
# ROC-AUC + PR-AUC EVALUATION
# ============================================================
# PURPOSE:
# Add publication-standard ranking metrics
#
# METRICS:
# - ROC-AUC
# - PR-AUC
# - Precision-Recall Curve
# - ROC Curve
#
# WORKS WITH:
# - best_sahajbert_gold
# - best_dapt_classifier
# - best_xlmr_model
# - best_muril_model
# - best_indicbert_model
# ============================================================

# ============================================================
# IMPORTS
# ============================================================

import os
import gc
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from datasets import Dataset

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer
)

import torch.nn.functional as F

# ============================================================
# SETTINGS
# ============================================================

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

print("DEVICE:", DEVICE)

# ============================================================
# MODEL PATHS
# ============================================================

MODELS = {

    'Baseline_SahajBERT':
        './best_sahajbert_gold',

    'DAPT_SahajBERT':
        './best_dapt_classifier',

    'MuRIL':
        './best_muril_model',

    'IndicBERTv2':
        './best_indicbert_model'
}

# ============================================================
# LOAD TEST DATA
# ============================================================

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

print("\nTest samples:", len(test_df))

# ============================================================
# HF DATASET
# ============================================================

test_ds = Dataset.from_pandas(test_df)

# ============================================================
# STORAGE
# ============================================================

results_rows = []

# ============================================================
# MAIN LOOP
# ============================================================

for model_name, model_path in MODELS.items():

    print("\n================================================")
    print("MODEL:", model_name)
    print("================================================")

    gc.collect()

    torch.cuda.empty_cache()

    # ========================================================
    # TOKENIZER
    # ========================================================

    tokenizer = AutoTokenizer.from_pretrained(
        model_path
    )

    # ========================================================
    # TOKENIZATION
    # ========================================================

    def tokenize(batch):

        return tokenizer(

            batch['text'],

            truncation=True,

            padding='max_length',

            max_length=128
        )

    tokenized_test = test_ds.map(
        tokenize,
        batched=True
    )

    tokenized_test.set_format(

        type='torch',

        columns=[
            'input_ids',
            'attention_mask',
            'label'
        ]
    )

    # ========================================================
    # MODEL
    # ========================================================

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path
    )

    model.to(DEVICE)

    # ========================================================
    # TRAINER
    # ========================================================

    trainer = Trainer(
        model=model
    )

    # ========================================================
    # PREDICTIONS
    # ========================================================

    predictions = trainer.predict(
        tokenized_test
    )

    logits = predictions.predictions

    y_true = predictions.label_ids

    probs = F.softmax(

        torch.tensor(logits),

        dim=1

    ).numpy()

    # probability of depressive class
    y_scores = probs[:, 1]

    # ========================================================
    # ROC-AUC
    # ========================================================

    roc_auc = roc_auc_score(
        y_true,
        y_scores
    )

    # ========================================================
    # PR-AUC
    # ========================================================

    pr_auc = average_precision_score(
        y_true,
        y_scores
    )

    print(f"\nROC-AUC: {roc_auc:.4f}")
    print(f"PR-AUC:  {pr_auc:.4f}")

    # ========================================================
    # SAVE RESULTS
    # ========================================================

    results_rows.append({

        'model':
            model_name,

        'roc_auc':
            roc_auc,

        'pr_auc':
            pr_auc
    })

    # ========================================================
    # ROC CURVE
    # ========================================================

    fpr, tpr, _ = roc_curve(
        y_true,
        y_scores
    )

    plt.figure(figsize=(6,6))

    plt.plot(
        fpr,
        tpr,
        label=f'AUC = {roc_auc:.4f}'
    )

    plt.plot(
        [0,1],
        [0,1],
        linestyle='--'
    )

    plt.xlabel('False Positive Rate')

    plt.ylabel('True Positive Rate')

    plt.title(f'ROC Curve — {model_name}')

    plt.legend()

    plt.tight_layout()

    plt.savefig(

        f'./BanglishDetect/results/{model_name}_roc_curve.png',

        dpi=300,

        bbox_inches='tight'
    )

    plt.show()

    # ========================================================
    # PR CURVE
    # ========================================================

    precision, recall, _ = precision_recall_curve(
        y_true,
        y_scores
    )

    plt.figure(figsize=(6,6))

    plt.plot(
        recall,
        precision,
        label=f'PR-AUC = {pr_auc:.4f}'
    )

    plt.xlabel('Recall')

    plt.ylabel('Precision')

    plt.title(f'Precision-Recall Curve — {model_name}')

    plt.legend()

    plt.tight_layout()

    plt.savefig(

        f'./BanglishDetect/results/{model_name}_pr_curve.png',

        dpi=300,

        bbox_inches='tight'
    )

    plt.show()

# ============================================================
# FINAL TABLE
# ============================================================

auc_df = pd.DataFrame(results_rows)

print("\n================================================")
print("FINAL AUC RESULTS")
print("================================================")

print(auc_df)

# ============================================================
# SAVE TABLE
# ============================================================

auc_df.to_csv(

    './BanglishDetect/results/auc_results.csv',

    index=False
)

print("\n✅ AUC analysis complete")
```

```python
# ============================================================
# MCNEMAR TEST
# ============================================================
# PURPOSE:
# Statistical significance test for paired classifiers
#
# COMPARES:
# - Baseline SahajBERT
# - DAPT-SahajBERT
#
# WHY IMPORTANT?
# McNemar is highly respected in NLP for:
# - same test set
# - paired predictions
# - classifier comparison
#
# OUTPUT:
# - contingency table
# - exact McNemar p-value
# - disagreement analysis
# - publication-ready interpretation
# ============================================================

# ============================================================
# IMPORTS
# ============================================================

import os
import gc
import torch
import numpy as np
import pandas as pd

from datasets import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer
)

from statsmodels.stats.contingency_tables import mcnemar

# ============================================================
# SETTINGS
# ============================================================

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

print("DEVICE:", DEVICE)

# ============================================================
# MODEL PATHS
# ============================================================

BASELINE_MODEL = './best_sahajbert_gold'

DAPT_MODEL = './best_dapt_classifier'

# ============================================================
# LOAD TEST DATA
# ============================================================

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

print("\nTest samples:", len(test_df))

# ============================================================
# HF DATASET
# ============================================================

test_ds = Dataset.from_pandas(test_df)

# ============================================================
# TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    DAPT_MODEL
)

# ============================================================
# TOKENIZATION
# ============================================================

def tokenize(batch):

    return tokenizer(

        batch['text'],

        truncation=True,

        padding='max_length',

        max_length=128
    )

test_ds = test_ds.map(
    tokenize,
    batched=True
)

test_ds.set_format(

    type='torch',

    columns=[
        'input_ids',
        'attention_mask',
        'label'
    ]
)

# ============================================================
# PREDICTION FUNCTION
# ============================================================

def get_predictions(model_path):

    print(f"\nLoading model: {model_path}")

    gc.collect()

    torch.cuda.empty_cache()

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path
    )

    model.to(DEVICE)

    trainer = Trainer(
        model=model
    )

    outputs = trainer.predict(test_ds)

    logits = outputs.predictions

    preds = np.argmax(logits, axis=1)

    labels = outputs.label_ids

    return preds, labels

# ============================================================
# GET PREDICTIONS
# ============================================================

baseline_preds, labels = get_predictions(
    BASELINE_MODEL
)

dapt_preds, _ = get_predictions(
    DAPT_MODEL
)

# ============================================================
# CORRECTNESS
# ============================================================

baseline_correct = (
    baseline_preds == labels
)

dapt_correct = (
    dapt_preds == labels
)

# ============================================================
# MCNEMAR CONTINGENCY TABLE
# ============================================================
#
#               DAPT
#             Correct Wrong
#
# Baseline
# Correct      a       b
# Wrong        c       d
#
# b = baseline correct, dapt wrong
# c = baseline wrong, dapt correct
#
# McNemar focuses on:
# disagreement pairs (b vs c)
# ============================================================

a = np.sum(
    (baseline_correct == True)
    &
    (dapt_correct == True)
)

b = np.sum(
    (baseline_correct == True)
    &
    (dapt_correct == False)
)

c = np.sum(
    (baseline_correct == False)
    &
    (dapt_correct == True)
)

d = np.sum(
    (baseline_correct == False)
    &
    (dapt_correct == False)
)

table = [

    [a, b],

    [c, d]
]

# ============================================================
# RUN MCNEMAR
# ============================================================

result = mcnemar(

    table,

    exact=True
)

# ============================================================
# PRINT RESULTS
# ============================================================

print("\n================================================")
print("MCNEMAR CONTINGENCY TABLE")
print("================================================")

print(pd.DataFrame(

    table,

    index=[
        'Baseline Correct',
        'Baseline Wrong'
    ],

    columns=[
        'DAPT Correct',
        'DAPT Wrong'
    ]
))

print("\n================================================")
print("MCNEMAR TEST RESULTS")
print("================================================")

print(f"Statistic: {result.statistic}")

print(f"P-value:   {result.pvalue:.8f}")

# ============================================================
# INTERPRETATION
# ============================================================

alpha = 0.05

print("\n================================================")
print("INTERPRETATION")
print("================================================")

if result.pvalue < alpha:

    print(
        "Result is statistically significant "
        "(p < 0.05)"
    )

    print(
        "DAPT-SahajBERT significantly differs "
        "from baseline."
    )

    if c > b:

        print(
            "\nDAPT corrected more errors "
            "than it introduced."
        )

        print(
            "This supports the effectiveness "
            "of domain-adaptive pretraining."
        )

else:

    print(
        "Difference is NOT statistically significant."
    )

# ============================================================
# SAVE RESULTS
# ============================================================

summary_df = pd.DataFrame({

    'metric': [

        'both_correct',
        'baseline_correct_dapt_wrong',
        'baseline_wrong_dapt_correct',
        'both_wrong',
        'mcnemar_statistic',
        'mcnemar_pvalue'
    ],

    'value': [

        a,
        b,
        c,
        d,
        result.statistic,
        result.pvalue
    ]
})

summary_df.to_csv(

    './BanglishDetect/results/mcnemar_results.csv',

    index=False
)

print("\n✅ McNemar analysis saved")
```

```python
# ============================================================
# COMPLETE ATTENTION + SHAP VISUALIZATION PIPELINE
# FINAL FIXED VERSION
# BANGLA + BANGLISH SUPPORT
# ============================================================

# ============================================================
# INSTALLS (RUN IF NEEDED)
# ============================================================

# !pip install transformers shap sentencepiece matplotlib pandas torch -q

# ============================================================
# IMPORTS
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
import shap
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

# ============================================================
# CONFIG
# ============================================================

MODEL_PATH = "./best_sahajbert_gold"

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

SAVE_DIR = "explainability_outputs"

os.makedirs(SAVE_DIR, exist_ok=True)

print("DEVICE:", DEVICE)

# ============================================================
# BANGLA FONT FIX
# ============================================================

BANGLA_FONT_PATH = (
    r"C:\Users\AFA001\AppData\Local\Microsoft\Windows\Fonts"
    r"\NotoSansBengali-VariableFont_wdth,wght.ttf"
)

if os.path.isfile(BANGLA_FONT_PATH):

    fm.fontManager.addfont(BANGLA_FONT_PATH)

    font_prop = fm.FontProperties(
        fname=BANGLA_FONT_PATH
    )

    plt.rcParams["font.family"] = font_prop.get_name()

    print("✅ Bangla font loaded")
    print("Font name:", font_prop.get_name())

else:

    print("⚠ Bangla font not found")
    print("Plots may show boxes for Bangla text")

# ============================================================
# LOAD TOKENIZER
# ============================================================

print("\nLoading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_PATH
)

# ============================================================
# LOAD MODEL
# IMPORTANT:
# attn_implementation='eager'
# REQUIRED FOR output_attentions=True
# ============================================================

print("\nLoading model...")

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_PATH,
    attn_implementation="eager"
)

model.to(DEVICE)

model.eval()

print("✅ Model loaded")

print("\nMODEL TYPE:")
print(type(model))

print("\nATTENTION IMPLEMENTATION:")
print(model.config._attn_implementation)

# ============================================================
# SAMPLE TEXTS
# ============================================================

texts = [

    "মিস করসি অতীত কে [SAD_FACE] [PLEADING]",

    "only broken heart can understand the pain [BROKEN_HEART]",

    "হেরে গেছি বাস্তবতার মঞ্চে",

    "নিজেদেরকে শান্তনা দেওয়ার আমরা বলি single paper can't decide future",

    "প্রিয় মানুষগুলোর খারাপ ব্যবহারে যখন নিজেকে একা লাগে"
]

# ============================================================
# TOKEN CLEANER
# ============================================================

def clean_tokens(tokens):

    cleaned = []

    for tok in tokens:

        tok = tok.replace("▁", "")

        tok = tok.strip()

        if tok == "":
            continue

        cleaned.append(tok)

    return cleaned

# ============================================================
# ATTENTION VISUALIZATION
# ============================================================

def visualize_attention(text, idx):

    print("\n" + "=" * 60)
    print("TEXT:")
    print(text)
    print("=" * 60)

    # ========================================================
    # TOKENIZATION
    # ========================================================

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    inputs = {
        k: v.to(DEVICE)
        for k, v in inputs.items()
    }

    # ========================================================
    # MODEL INFERENCE
    # ========================================================

    with torch.no_grad():

        outputs = model(
            **inputs,
            output_attentions=True,
            return_dict=True
        )

    # ========================================================
    # GET ATTENTIONS
    # ========================================================

    attentions = outputs.attentions

    if attentions is None or len(attentions) == 0:

        print("⚠ No attentions returned")

        if hasattr(outputs, "keys"):

            print("\nAvailable output keys:")
            print(outputs.keys())

        return

    # ========================================================
    # LAST LAYER ATTENTION
    # ========================================================

    last_attention = attentions[-1]

    # shape:
    # [batch, heads, seq_len, seq_len]

    attention_scores = last_attention.mean(dim=1)

    attention_scores = attention_scores[0]

    # CLS attention
    cls_attention = attention_scores[0]

    cls_attention = cls_attention.detach().cpu().numpy()

    # ========================================================
    # TOKENS
    # ========================================================

    input_ids = inputs["input_ids"][0]

    tokens = tokenizer.convert_ids_to_tokens(
        input_ids.detach().cpu().numpy()
    )

    # ========================================================
    # FILTER TOKENS
    # ========================================================

    filtered_tokens = []

    filtered_scores = []

    for tok, score in zip(tokens, cls_attention):

        if tok in tokenizer.all_special_tokens:
            continue

        tok = tok.replace("▁", "")

        tok = tok.strip()

        if tok == "":
            continue

        filtered_tokens.append(tok)

        filtered_scores.append(float(score))

    # ========================================================
    # DATAFRAME
    # ========================================================

    attention_df = pd.DataFrame({

        "token": filtered_tokens,
        "attention_score": filtered_scores
    })

    attention_df = attention_df.sort_values(

        by="attention_score",
        ascending=False
    )

    print("\nTOP ATTENDED TOKENS:\n")

    print(attention_df.head(10))

    # ========================================================
    # PLOT
    # ========================================================

    plt.figure(figsize=(12, 6))

    plot_tokens = filtered_tokens[:20]

    plot_scores = filtered_scores[:20]

    plt.bar(
        range(len(plot_tokens)),
        plot_scores
    )

    plt.xticks(
        range(len(plot_tokens)),
        plot_tokens,
        rotation=45,
        fontsize=11
    )

    plt.ylabel("Attention Score")

    plt.title("Attention Visualization")

    plt.tight_layout()

    save_path = os.path.join(
        SAVE_DIR,
        f"attention_{idx}.png"
    )

    plt.savefig(
        save_path,
        dpi=300
    )

    plt.show()

    plt.close()

    print(f"\n✅ Saved: {save_path}")

# ============================================================
# RUN ATTENTION VISUALIZATION
# ============================================================

print("\n" + "=" * 60)
print("RUNNING ATTENTION VISUALIZATION")
print("=" * 60)

for i, text in enumerate(texts):

    visualize_attention(text, i)

# ============================================================
# SHAP PREDICTION FUNCTION
# ============================================================

def predict_proba(texts_input):

    # ========================================================
    # FIX SHAP INPUT TYPES
    # ========================================================

    if isinstance(texts_input, np.ndarray):

        texts_input = texts_input.tolist()

    if isinstance(texts_input, str):

        texts_input = [texts_input]

    texts_input = [str(x) for x in texts_input]

    # ========================================================
    # TOKENIZATION
    # ========================================================

    inputs = tokenizer(
        texts_input,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    )

    inputs = {
        k: v.to(DEVICE)
        for k, v in inputs.items()
    }

    # ========================================================
    # INFERENCE
    # ========================================================

    with torch.no_grad():

        outputs = model(**inputs)

    logits = outputs.logits

    probs = torch.softmax(
        logits,
        dim=1
    )

    return probs.detach().cpu().numpy()

# ============================================================
# SHAP EXPLAINER
# ============================================================

print("\n" + "=" * 60)
print("RUNNING SHAP")
print("=" * 60)

explainer = shap.Explainer(
    predict_proba,
    tokenizer
)

# ============================================================
# COMPUTE SHAP VALUES
# ============================================================

shap_values = explainer(
    texts,
    fixed_context=1
)

print("✅ SHAP values computed")

# ============================================================
# SHAP TEXT VISUALIZATION
# ============================================================

for i in range(len(texts)):

    print("\n" + "=" * 60)
    print("SHAP FOR:")
    print(texts[i])
    print("=" * 60)

    shap.plots.text(
        shap_values[i],
        display=False
    )

# ============================================================
# SAVE SHAP BAR PLOTS
# ============================================================

try:

    for i in range(len(texts)):

        plt.figure(figsize=(10, 5))

        shap.plots.bar(
            shap_values[i][:, :, 1],
            show=False
        )

        save_path = os.path.join(
            SAVE_DIR,
            f"shap_bar_{i}.png"
        )

        plt.tight_layout()

        plt.savefig(
            save_path,
            dpi=300
        )

        plt.close()

        print(f"✅ Saved SHAP bar plot: {save_path}")

except Exception as e:

    print("\n⚠ Could not save SHAP bar plots")

    print(e)

# ============================================================
# PREDICTIONS
# ============================================================

print("\n" + "=" * 60)
print("MODEL PREDICTIONS")
print("=" * 60)

preds = predict_proba(texts)

for text, prob in zip(texts, preds):

    pred_class = np.argmax(prob)

    confidence = np.max(prob)

    print("\nTEXT:")
    print(text)

    print(f"PREDICTED CLASS: {pred_class}")

    print(f"CONFIDENCE: {confidence:.4f}")

    print(f"PROBABILITIES: {prob}")

# ============================================================
# DONE
# ============================================================

print("\n" + "=" * 60)
print("ALL DONE")
print("=" * 60)

print(f"Outputs saved in: {SAVE_DIR}")
```

```python
# Final Banglish Depression Detection Inference Pipeline

# ============================================================
# FINAL BANGLISH DEPRESSION DETECTION PIPELINE
# ============================================================
# Author: Ayon Roy
# Project: BanglishDepYT + EPH Preprocessing
# Model: SahajBERT Depression Classifier
# ============================================================

# =========================
# IMPORTS
# =========================

import os
import re
import torch
import pandas as pd
import numpy as np
from tqdm import tqdm

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

# =========================
# DEVICE
# =========================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("DEVICE:", DEVICE)

# =========================
# MODEL PATH
# =========================

MODEL_PATH = "./best_sahajbert_gold"

# =========================
# LOAD MODEL + TOKENIZER
# =========================

print("\nLoading model...")

try:
    tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_PATH
    )

    model.to(DEVICE)
    model.eval()

    print("Model loaded successfully!")

except Exception as e:
    print("Error loading model:")
    print(e)
    raise

# =========================
# CONFIG
# =========================

MAX_LEN = 128
BATCH_SIZE = 32

# =========================
# EPH PREPROCESSING
# =========================
# Emotion Preserving Hybrid (EPH)
# Keeps:
# - Emojis
# - Emoticons
# - Banglish
# - Repetitions
# - Emotional punctuation
# Removes:
# - URLs
# - Mentions
# - Extra spaces
#
# This is your final P4 preprocessing.
# =========================

EMOTICON_PATTERN = r'(:\)|:\(|:D|;-\)|<3|:\'\(|T_T|xD|XD|D:|:\||=\)|=\(|:\]|:\[|:\*)'


def clean_text_eph(text):
    """
    Final EPH preprocessing pipeline.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # Remove URLs
    text = re.sub(r'http\S+|www\.\S+', ' ', text)

    # Remove mentions
    text = re.sub(r'@\w+', ' ', text)

    # Remove hashtags symbol only
    text = re.sub(r'#', '', text)

    # Preserve emojis + Banglish + emotion
    # Remove weird control chars
    text = re.sub(r'[\r\n\t]', ' ', text)

    # Normalize spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# =========================
# SINGLE PREDICTION FUNCTION
# =========================


def predict_single(text):
    """
    Predict a single sentence.
    """

    cleaned_text = clean_text_eph(text)

    encoding = tokenizer(
        cleaned_text,
        truncation=True,
        padding='max_length',
        max_length=MAX_LEN,
        return_tensors='pt'
    )

    input_ids = encoding['input_ids'].to(DEVICE)
    attention_mask = encoding['attention_mask'].to(DEVICE)

    with torch.no_grad():
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        logits = outputs.logits

        probs = torch.softmax(logits, dim=1)

        pred_class = torch.argmax(probs, dim=1).item()

        confidence = probs[0][pred_class].item()

    label_map = {
        0: 'Non-Depressive',
        1: 'Depressive'
    }

    return {
        'original_text': text,
        'cleaned_text': cleaned_text,
        'prediction': label_map[pred_class],
        'confidence': round(confidence, 4)
    }

# =========================
# BATCH PREDICTION
# =========================


def predict_batch(texts):
    """
    Predict multiple texts.
    """

    cleaned_texts = [clean_text_eph(t) for t in texts]

    encodings = tokenizer(
        cleaned_texts,
        truncation=True,
        padding=True,
        max_length=MAX_LEN,
        return_tensors='pt'
    )

    input_ids = encodings['input_ids'].to(DEVICE)
    attention_mask = encodings['attention_mask'].to(DEVICE)

    with torch.no_grad():
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        logits = outputs.logits

        probs = torch.softmax(logits, dim=1)

        predictions = torch.argmax(probs, dim=1).cpu().numpy()

        confidences = torch.max(probs, dim=1).values.cpu().numpy()

    label_map = {
        0: 'Non-Depressive',
        1: 'Depressive'
    }

    results = []

    for i in range(len(texts)):
        results.append({
            'text': texts[i],
            'cleaned_text': cleaned_texts[i],
            'prediction': label_map[predictions[i]],
            'confidence': round(float(confidences[i]), 4)
        })

    return pd.DataFrame(results)

# =========================
# CSV INFERENCE PIPELINE
# =========================


def predict_csv(
    input_csv,
    text_column='text',
    output_csv='predictions.csv'
):
    """
    Run prediction on a CSV file.
    """

    print(f"\nLoading CSV: {input_csv}")

    df = pd.read_csv(input_csv)

    print(f"Total rows: {len(df)}")

    predictions = []
    confidences = []
    cleaned_texts = []

    texts = df[text_column].astype(str).tolist()

    for i in tqdm(range(0, len(texts), BATCH_SIZE)):

        batch_texts = texts[i:i + BATCH_SIZE]

        batch_cleaned = [clean_text_eph(t) for t in batch_texts]

        encodings = tokenizer(
            batch_cleaned,
            truncation=True,
            padding=True,
            max_length=MAX_LEN,
            return_tensors='pt'
        )

        input_ids = encodings['input_ids'].to(DEVICE)
        attention_mask = encodings['attention_mask'].to(DEVICE)

        with torch.no_grad():
            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            logits = outputs.logits

            probs = torch.softmax(logits, dim=1)

            batch_preds = torch.argmax(probs, dim=1).cpu().numpy()

            batch_conf = torch.max(
                probs,
                dim=1
            ).values.cpu().numpy()

        predictions.extend(batch_preds)
        confidences.extend(batch_conf)
        cleaned_texts.extend(batch_cleaned)

    label_map = {
        0: 'Non-Depressive',
        1: 'Depressive'
    }

    df['cleaned_text'] = cleaned_texts
    df['prediction_id'] = predictions
    df['prediction'] = [label_map[p] for p in predictions]
    df['confidence'] = np.round(confidences, 4)

    df.to_csv(output_csv, index=False)

    print(f"\nPredictions saved to: {output_csv}")

    print("\nPrediction Distribution:")
    print(df['prediction'].value_counts())

    return df

# ============================================================
# TEST SINGLE SAMPLE
# ============================================================

print("\n========================")
print("SINGLE SAMPLE TEST")
print("========================")

sample_texts = [
    "আমার আর বাঁচতে ইচ্ছা করে না 😭",
    "আজকে অনেক ভালো লাগছে আলহামদুলিল্লাহ ❤️",
    "I feel so depressed and alone",
    "সবাই আমাকে ছেড়ে চলে গেছে",
    "life is beautiful :)"
]

for text in sample_texts:

    result = predict_single(text)

    print("\nTEXT:")
    print(result['original_text'])

    print("PREDICTION:", result['prediction'])
    print("CONFIDENCE:", result['confidence'])

# ============================================================
# EXAMPLE CSV USAGE
# ============================================================

# Uncomment to use

# df_result = predict_csv(
#     input_csv='youtube_comments.csv',
#     text_column='text',
#     output_csv='youtube_predictions.csv'
# )

# ============================================================
# HIGH-CONFIDENCE PSEUDO LABELING
# ============================================================
# Useful for semi-supervised expansion.
#
# Example:
# Keep only predictions with confidence >= 0.90
# ============================================================


def generate_pseudo_labels(
    input_csv,
    text_column='text',
    confidence_threshold=0.90,
    output_csv='pseudo_labeled.csv'
):

    df = predict_csv(
        input_csv=input_csv,
        text_column=text_column,
        output_csv='temp_predictions.csv'
    )

    high_conf = df[
        df['confidence'] >= confidence_threshold
    ].copy()

    high_conf.to_csv(output_csv, index=False)

    print("\nPseudo-labeling completed!")
    print(f"Threshold: {confidence_threshold}")
    print(f"Selected samples: {len(high_conf)}")

    return high_conf

# ============================================================
# SAVE CLEAN HUGGINGFACE MODEL
# ============================================================

# Optional:
# Save final clean version for deployment

# model.save_pretrained("final_sahajbert_eph_model")
# tokenizer.save_pretrained("final_sahajbert_eph_model")

print("\nSystem ready for inference!")
```
