```python
# ============================================================
# FINAL DAPT CLASSIFIER TRAINING + SAVE
# ============================================================
# PURPOSE:
# Train final depression classifier using:
# DAPT-SahajBERT encoder
#
# OUTPUT:
# ./best_dapt_classifier
#
# IMPORTANT:
# This is the FINAL classifier checkpoint
# used for:
# - multi-seed evaluation
# - error analysis
# - paper results
# ============================================================

import os
import gc
import random
import numpy as np
import pandas as pd
import torch

from datasets import Dataset

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    matthews_corrcoef
)

import transformers.modeling_utils as mu
mu.check_torch_load_is_safe = lambda: None

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback
)

# ============================================================
# SETTINGS
# ============================================================

SEED = 42

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

print("DEVICE:", DEVICE)

os.environ["TOKENIZERS_PARALLELISM"] = "false"

# ============================================================
# PATHS
# ============================================================

# DAPT MLM checkpoint
DAPT_MODEL_PATH = './dapt_sahajbert'

# Final classifier save path
FINAL_SAVE_PATH = './best_dapt_classifier'

# ============================================================
# LOAD GOLD SPLITS
# ============================================================

train_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_train.csv'
)

val_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_val.csv'
)

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

print("\nDATA:")
print("Train:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))

# ============================================================
# TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    DAPT_MODEL_PATH
)

# ============================================================
# HF DATASETS
# ============================================================

train_ds = Dataset.from_pandas(train_df)

val_ds = Dataset.from_pandas(val_df)

test_ds = Dataset.from_pandas(test_df)

# ============================================================
# TOKENIZATION
# ============================================================

def tokenize(batch):

    return tokenizer(

        batch['text'],

        truncation=True,

        padding='max_length',

        max_length=128
    )

print("\nTokenizing datasets...")

train_ds = train_ds.map(
    tokenize,
    batched=True
)

val_ds = val_ds.map(
    tokenize,
    batched=True
)

test_ds = test_ds.map(
    tokenize,
    batched=True
)

cols = [
    'input_ids',
    'attention_mask',
    'label'
]

train_ds.set_format(
    type='torch',
    columns=cols
)

val_ds.set_format(
    type='torch',
    columns=cols
)

test_ds.set_format(
    type='torch',
    columns=cols
)

print("✅ Tokenization complete")

# ============================================================
# LOAD DAPT MODEL AS CLASSIFIER
# ============================================================

print("\nLoading DAPT model...")

model = AutoModelForSequenceClassification.from_pretrained(

    DAPT_MODEL_PATH,

    num_labels=2,

    ignore_mismatched_sizes=True
)

model.to(DEVICE)

print("✅ DAPT classifier loaded")

# ============================================================
# METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    preds = np.argmax(logits, axis=1)

    return {

        'accuracy':
            accuracy_score(labels, preds),

        'f1':
            f1_score(labels, preds),

        'precision':
            precision_score(labels, preds),

        'recall':
            recall_score(labels, preds),

        'mcc':
            matthews_corrcoef(labels, preds)
    }

# ============================================================
# TRAINING ARGS
# ============================================================

training_args = TrainingArguments(

    output_dir='./tmp_dapt_classifier',

    eval_strategy='epoch',

    save_strategy='epoch',

    logging_strategy='epoch',

    learning_rate=1e-5,

    per_device_train_batch_size=4,

    per_device_eval_batch_size=4,

    gradient_accumulation_steps=4,

    num_train_epochs=8,

    weight_decay=0.01,

    warmup_steps=100,

    fp16=False,

    dataloader_num_workers=0,

    load_best_model_at_end=True,

    metric_for_best_model='f1',

    greater_is_better=True,

    save_total_limit=1,

    seed=SEED,

    report_to='none'
)

# ============================================================
# TRAINER
# ============================================================

trainer = Trainer(

    model=model,

    args=training_args,

    train_dataset=train_ds,

    eval_dataset=val_ds,

    compute_metrics=compute_metrics,

    callbacks=[

        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)

# ============================================================
# TRAIN
# ============================================================

gc.collect()

torch.cuda.empty_cache()

print("\n================================================")
print("TRAINING FINAL DAPT CLASSIFIER")
print("================================================")

trainer.train()

# ============================================================
# EVALUATE
# ============================================================

print("\n================================================")
print("EVALUATING ON GOLD TEST SET")
print("================================================")

results = trainer.evaluate(test_ds)

print("\nFINAL DAPT RESULTS:")

for k, v in results.items():

    print(f"{k}: {v}")

# ============================================================
# SAVE FINAL MODEL
# ============================================================

print("\nSaving final DAPT classifier...")

trainer.save_model(FINAL_SAVE_PATH)

tokenizer.save_pretrained(
    FINAL_SAVE_PATH
)

print("\n✅ FINAL DAPT CLASSIFIER SAVED")

# ============================================================
# VERIFY FILES
# ============================================================

print("\nSaved files:")

print(
    os.listdir(FINAL_SAVE_PATH)
)

# ============================================================
# CLEANUP
# ============================================================

gc.collect()

torch.cuda.empty_cache()

print("\n✅ COMPLETE")
```

```python
# ============================================================
# FULL STATISTICAL ANALYSIS
# ============================================================
# PURPOSE:
# Load multi-seed experiment results
# and compute:
#
# - mean ± std
# - paired t-test
# - Cohen's d
#
# OUTPUT:
# publication-ready statistical table
# ============================================================

import numpy as np
import pandas as pd

from scipy.stats import ttest_rel

# ============================================================
# LOAD RESULTS
# ============================================================

results_df = pd.read_csv(
    './multi_seed_results.csv'
)

print("\n================================================")
print("LOADED RESULTS")
print("================================================")

print(results_df.head())

# ============================================================
# SPLIT RESULTS
# ============================================================

baseline_df = results_df[
    results_df['model'] == 'baseline'
].copy()

dapt_df = results_df[
    results_df['model'] == 'dapt'
].copy()

# ============================================================
# SORT BY SEED
# IMPORTANT FOR PAIRED TESTS
# ============================================================

baseline_df = baseline_df.sort_values(
    by='seed'
).reset_index(drop=True)

dapt_df = dapt_df.sort_values(
    by='seed'
).reset_index(drop=True)

print("\nBaseline rows:", len(baseline_df))
print("DAPT rows:", len(dapt_df))

# ============================================================
# VERIFY SAME SEEDS
# ============================================================

assert list(baseline_df['seed']) == list(dapt_df['seed']), \
    "Seed mismatch between baseline and DAPT"

print("\n✅ Seeds aligned correctly")

# ============================================================
# METRICS
# ============================================================

metrics = [
    'accuracy',
    'f1',
    'precision',
    'recall',
    'mcc'
]

# ============================================================
# COHEN'S D (PAIRED)
# ============================================================

def cohens_d_paired(x, y):

    diff = np.array(y) - np.array(x)

    std_diff = diff.std(ddof=1)

    if std_diff == 0:
        return 0.0

    return diff.mean() / std_diff

# ============================================================
# ANALYSIS
# ============================================================

final_rows = []

for metric in metrics:

    print("\n------------------------------------------------")
    print("METRIC:", metric)
    print("------------------------------------------------")

    # ========================================================
    # SCORES
    # ========================================================

    base_scores = baseline_df[metric].values

    dapt_scores = dapt_df[metric].values

    print("Baseline:", base_scores)
    print("DAPT:    ", dapt_scores)

    # ========================================================
    # MEAN ± STD
    # ========================================================

    base_mean = np.mean(base_scores)

    base_std = np.std(
        base_scores,
        ddof=1
    )

    dapt_mean = np.mean(dapt_scores)

    dapt_std = np.std(
        dapt_scores,
        ddof=1
    )

    # ========================================================
    # PAIRED T-TEST
    # ========================================================

    t_stat, p_value = ttest_rel(
        dapt_scores,
        base_scores
    )

    # ========================================================
    # COHEN'S D
    # ========================================================

    d_value = cohens_d_paired(
        base_scores,
        dapt_scores
    )

    # ========================================================
    # SAVE ROW
    # ========================================================

    row = {

        'metric':
            metric,

        'baseline_mean':
            round(base_mean, 4),

        'baseline_std':
            round(base_std, 4),

        'dapt_mean':
            round(dapt_mean, 4),

        'dapt_std':
            round(dapt_std, 4),

        'p_value':
            round(p_value, 6),

        'cohens_d':
            round(d_value, 4)
    }

    final_rows.append(row)

# ============================================================
# FINAL DATAFRAME
# ============================================================

stats_df = pd.DataFrame(final_rows)

# ============================================================
# PRETTY PAPER TABLE
# ============================================================

paper_table = pd.DataFrame({

    'Metric':
        stats_df['metric'],

    'Baseline':
        stats_df.apply(
            lambda x:
            f"{x['baseline_mean']:.4f} ± {x['baseline_std']:.4f}",
            axis=1
        ),

    'DAPT':
        stats_df.apply(
            lambda x:
            f"{x['dapt_mean']:.4f} ± {x['dapt_std']:.4f}",
            axis=1
        ),

    'p-value':
        stats_df['p_value'],

    "Cohen's d":
        stats_df['cohens_d']
})

# ============================================================
# PRINT RESULTS
# ============================================================

print("\n================================================")
print("STATISTICAL ANALYSIS")
print("================================================")

print(stats_df)

print("\n================================================")
print("PAPER TABLE")
print("================================================")

print(paper_table)

# ============================================================
# SAVE
# ============================================================

stats_df.to_csv(
    './statistical_analysis.csv',
    index=False
)

paper_table.to_csv(
    './paper_ready_stats_table.csv',
    index=False
)

print("\n✅ statistical_analysis.csv saved")

print("✅ paper_ready_stats_table.csv saved")

# ============================================================
# SIGNIFICANCE INTERPRETATION
# ============================================================

print("\n================================================")
print("INTERPRETATION")
print("================================================")

for _, row in stats_df.iterrows():

    metric = row['metric']

    p = row['p_value']

    d = row['cohens_d']

    if p < 0.05:
        significance = "SIGNIFICANT"
    else:
        significance = "NOT significant"

    if abs(d) < 0.2:
        effect = "small"
    elif abs(d) < 0.5:
        effect = "medium"
    elif abs(d) < 0.8:
        effect = "large"
    else:
        effect = "very large"

    print(
        f"{metric}: "
        f"{significance}, "
        f"Cohen's d={d:.4f} ({effect} effect)"
    )

print("\n✅ COMPLETE")
```

```python
import pandas as pd
import numpy as np
import torch
import gc

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

# ============================================================
# DEVICE
# ============================================================

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

print("DEVICE:", DEVICE)

# ============================================================
# LOAD GOLD TEST SET
# ============================================================

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

texts = test_df['text'].astype(str).tolist()

labels = test_df['label'].tolist()

print(f"Test samples: {len(test_df)}")

# ============================================================
# IMPORTANT:
# USE ACTUAL TRAINED CLASSIFIERS
#
# baseline:
#   ./best_sahajbert_gold
#
# DAPT classifier:
#   ./dapt_classifier_results
#
# DO NOT USE:
#   ./dapt_sahajbert
#
# because that is MLM-only DAPT weights
# not fine-tuned classifier
# ============================================================

BASELINE_MODEL = './best_sahajbert_gold'

DAPT_MODEL = './best_dapt_classifier'

# ============================================================
# PREDICTION FUNCTION
# ============================================================

def predict_model(model_path):

    print(f"\nLoading model: {model_path}")

    # ====================================================
    # ALWAYS USE SAME TOKENIZER
    # ====================================================

    tokenizer = AutoTokenizer.from_pretrained(
        './best_sahajbert_gold'
    )

    # ====================================================
    # LOAD MODEL
    # ====================================================

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path,
        num_labels=2
    ).to(DEVICE)

    model.eval()

    preds = []

    probs_all = []

    BATCH = 16

    for i in range(0, len(texts), BATCH):

        batch = texts[i:i+BATCH]

        enc = tokenizer(

            batch,

            truncation=True,

            padding=True,

            max_length=128,

            return_tensors='pt'
        )

        enc = {
            k:v.to(DEVICE)
            for k,v in enc.items()
        }

        with torch.no_grad():

            outputs = model(**enc)

            logits = outputs.logits

            probs = torch.softmax(
                logits,
                dim=1
            )

            pred = torch.argmax(
                probs,
                dim=1
            )

            conf = torch.max(
                probs,
                dim=1
            ).values

        preds.extend(
            pred.cpu().numpy()
        )

        probs_all.extend(
            conf.cpu().numpy()
        )

    del model
    gc.collect()
    torch.cuda.empty_cache()

    return preds, probs_all

# ============================================================
# GET PREDICTIONS
# ============================================================

baseline_preds, baseline_conf = predict_model(
    BASELINE_MODEL
)

dapt_preds, dapt_conf = predict_model(
    DAPT_MODEL
)

# ============================================================
# BUILD ANALYSIS DATAFRAME
# ============================================================

analysis_df = pd.DataFrame({

    'text': texts,

    'true_label': labels,

    'baseline_pred': baseline_preds,

    'baseline_confidence': baseline_conf,

    'dapt_pred': dapt_preds,

    'dapt_confidence': dapt_conf
})

# ============================================================
# CATEGORIZE RESULTS
# ============================================================

def categorize(row):

    b_correct = (
        row['baseline_pred']
        ==
        row['true_label']
    )

    d_correct = (
        row['dapt_pred']
        ==
        row['true_label']
    )

    if b_correct and d_correct:

        return 'both_correct'

    elif (not b_correct) and d_correct:

        return 'baseline_wrong_dapt_correct'

    elif b_correct and (not d_correct):

        return 'baseline_correct_dapt_wrong'

    else:

        return 'both_wrong'

analysis_df['category'] = analysis_df.apply(
    categorize,
    axis=1
)

# ============================================================
# SUMMARY
# ============================================================

print("\n================================================")
print("ERROR ANALYSIS SUMMARY")
print("================================================")

print(
    analysis_df['category'].value_counts()
)

# ============================================================
# SAVE FULL FILE
# ============================================================

analysis_df.to_csv(
    './error_analysis.csv',
    index=False
)

print("\n✅ Full error analysis saved")

# ============================================================
# SAVE MOST IMPORTANT CASES
# ============================================================

improved_df = analysis_df[

    analysis_df['category']
    ==
    'baseline_wrong_dapt_correct'

].copy()

improved_df.to_csv(
    './dapt_improvement_cases.csv',
    index=False
)

print(
    f"\n✅ DAPT improvement cases: {len(improved_df)}"
)

# ============================================================
# SAVE REGRESSION CASES
# ============================================================

regression_df = analysis_df[

    analysis_df['category']
    ==
    'baseline_correct_dapt_wrong'

].copy()

regression_df.to_csv(
    './dapt_regression_cases.csv',
    index=False
)

print(
    f"✅ DAPT regression cases: {len(regression_df)}"
)

# ============================================================
# SHOW SAMPLE IMPROVEMENTS
# ============================================================

print("\n================================================")
print("SAMPLE DAPT IMPROVEMENTS")
print("================================================")

sample_improved = improved_df.head(10)

for idx, row in sample_improved.iterrows():

    print("\n--------------------------------------")

    print("TEXT:")
    print(row['text'])

    print(f"\nTRUE:      {row['true_label']}")

    print(
        f"BASELINE:  {row['baseline_pred']}"
    )

    print(
        f"DAPT:      {row['dapt_pred']}"
    )

    print(
        f"DAPT CONF: {row['dapt_confidence']:.4f}"
    )

print("\n✅ Error analysis complete")
```

```python
# ============================================================
# XLM-ROBERTA EXPERIMENT — FULL PIPELINE
# ============================================================
# FEATURES:
# - Uses frozen gold split
# - EPH preprocessed data
# - Proper evaluation
# - Multi-metric reporting
# - Confusion matrix
# - Classification report
# - Confidence histogram
# - Error analysis CSV
# - Publication-quality visualizations
# ============================================================

# ============================================================
# CELL 1 — IMPORTS
# ============================================================

import os
import gc
import torch
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from datasets import Dataset

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    matthews_corrcoef,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback
)

import torch.nn.functional as F

import transformers.modeling_utils as mu
mu.check_torch_load_is_safe = lambda: None

# ============================================================
# SETTINGS
# ============================================================

SEED = 42

MODEL_NAME = 'xlm-roberta-base'

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

RESULT_DIR = './BanglishDetect/results/xlmr/'

MODEL_SAVE_DIR = './best_xlmr_model'

os.makedirs(RESULT_DIR, exist_ok=True)

print("DEVICE:", DEVICE)

# ============================================================
# SEED
# ============================================================

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)

# ============================================================
# CELL 2 — LOAD DATA
# ============================================================

train_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_train.csv'
)

val_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_val.csv'
)

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

print("\nTrain:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))

# ============================================================
# CELL 3 — TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

# ============================================================
# HF DATASETS
# ============================================================

train_ds = Dataset.from_pandas(train_df)

val_ds = Dataset.from_pandas(val_df)

test_ds = Dataset.from_pandas(test_df)

# ============================================================
# TOKENIZATION
# ============================================================

def tokenize(batch):

    return tokenizer(

        batch['text'],

        truncation=True,

        padding='max_length',

        max_length=128
    )

train_ds = train_ds.map(
    tokenize,
    batched=True
)

val_ds = val_ds.map(
    tokenize,
    batched=True
)

test_ds = test_ds.map(
    tokenize,
    batched=True
)

cols = [
    'input_ids',
    'attention_mask',
    'label'
]

train_ds.set_format(
    type='torch',
    columns=cols
)

val_ds.set_format(
    type='torch',
    columns=cols
)

test_ds.set_format(
    type='torch',
    columns=cols
)

print("\n✅ Tokenization complete")

# ============================================================
# CELL 4 — LOAD MODEL
# ============================================================

model = AutoModelForSequenceClassification.from_pretrained(

    MODEL_NAME,

    num_labels=2
)

model.to(DEVICE)

print("\n✅ XLM-R loaded")

# ============================================================
# CELL 5 — METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    preds = np.argmax(logits, axis=1)

    return {

        'accuracy':
            accuracy_score(labels, preds),

        'f1':
            f1_score(labels, preds),

        'precision':
            precision_score(labels, preds),

        'recall':
            recall_score(labels, preds),

        'mcc':
            matthews_corrcoef(labels, preds)
    }

# ============================================================
# CELL 6 — TRAINING ARGUMENTS
# ============================================================

training_args = TrainingArguments(

    output_dir='./xlmr_results',

    eval_strategy='epoch',

    save_strategy='epoch',

    logging_strategy='epoch',

    learning_rate=2e-5,

    per_device_train_batch_size=8,

    per_device_eval_batch_size=8,

    gradient_accumulation_steps=2,

    num_train_epochs=5,

    warmup_steps=100,

    weight_decay=0.01,

    fp16=True,

    dataloader_num_workers=0,

    load_best_model_at_end=True,

    metric_for_best_model='f1',

    greater_is_better=True,

    save_total_limit=1,

    seed=SEED,

    report_to='none'
)

print("\n✅ Training arguments ready")

# ============================================================
# CELL 7 — TRAINER
# ============================================================

trainer = Trainer(

    model=model,

    args=training_args,

    train_dataset=train_ds,

    eval_dataset=val_ds,

    compute_metrics=compute_metrics,

    callbacks=[

        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)

print("\n✅ Trainer ready")

# ============================================================
# CELL 8 — TRAIN
# ============================================================

gc.collect()

torch.cuda.empty_cache()

print("\n================================================")
print("TRAINING XLM-R")
print("================================================")

trainer.train()

print("\n✅ Training complete")

# ============================================================
# CELL 9 — EVALUATION
# ============================================================

print("\n================================================")
print("TEST EVALUATION")
print("================================================")

results = trainer.evaluate(test_ds)

for k, v in results.items():

    print(f"{k}: {v}")

# ============================================================
# SAVE RESULTS
# ============================================================

results_df = pd.DataFrame([results])

results_df.to_csv(

    f'{RESULT_DIR}xlmr_test_results.csv',

    index=False
)

print("\n✅ Results saved")

# ============================================================
# CELL 10 — SAVE MODEL
# ============================================================

trainer.save_model(MODEL_SAVE_DIR)

tokenizer.save_pretrained(MODEL_SAVE_DIR)

print("\n✅ Model saved")

# ============================================================
# CELL 11 — PREDICTIONS
# ============================================================

pred_output = trainer.predict(test_ds)

logits = pred_output.predictions

y_true = pred_output.label_ids

y_pred = np.argmax(logits, axis=1)

probs = F.softmax(
    torch.tensor(logits),
    dim=1
).numpy()

confidences = probs.max(axis=1)

# ============================================================
# CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_true,

    y_pred,

    digits=4
)

print("\n================================================")
print("CLASSIFICATION REPORT")
print("================================================")

print(report)

with open(

    f'{RESULT_DIR}classification_report.txt',

    'w',

    encoding='utf-8'

) as f:

    f.write(report)

# ============================================================
# CELL 12 — CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

fig, ax = plt.subplots(figsize=(5,5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot(ax=ax)

plt.title('XLM-R Confusion Matrix')

plt.tight_layout()

plt.savefig(

    f'{RESULT_DIR}confusion_matrix.png',

    dpi=300,

    bbox_inches='tight'
)

plt.show()

# ============================================================
# CELL 13 — CONFIDENCE HISTOGRAM
# ============================================================

plt.figure(figsize=(8,5))

plt.hist(
    confidences,
    bins=20
)

plt.xlabel('Confidence')

plt.ylabel('Frequency')

plt.title('XLM-R Confidence Distribution')

plt.tight_layout()

plt.savefig(

    f'{RESULT_DIR}confidence_distribution.png',

    dpi=300,

    bbox_inches='tight'
)

plt.show()

# ============================================================
# CELL 14 — ERROR ANALYSIS
# ============================================================

analysis_df = pd.DataFrame({

    'text':
        test_df['text'],

    'true_label':
        y_true,

    'prediction':
        y_pred,

    'confidence':
        confidences
})

analysis_df['correct'] = (

    analysis_df['true_label']
    ==
    analysis_df['prediction']
)

# ============================================================
# SAVE FULL ANALYSIS
# ============================================================

analysis_df.to_csv(

    f'{RESULT_DIR}full_error_analysis.csv',

    index=False
)

print("\n✅ Full error analysis saved")

# ============================================================
# WRONG PREDICTIONS
# ============================================================

wrong_df = analysis_df[
    analysis_df['correct'] == False
]

wrong_df.to_csv(

    f'{RESULT_DIR}wrong_predictions.csv',

    index=False
)

print("✅ Wrong predictions saved")

# ============================================================
# SAMPLE ERRORS
# ============================================================

print("\n================================================")
print("SAMPLE ERRORS")
print("================================================")

for _, row in wrong_df.head(10).iterrows():

    print("\n--------------------------------------")

    print("TEXT:")
    print(row['text'])

    print("\nTRUE:", row['true_label'])

    print("PRED:", row['prediction'])

    print("CONF:", round(row['confidence'], 4))

# ============================================================
# CELL 15 — FINAL SUMMARY
# ============================================================

summary_df = pd.DataFrame({

    'Metric': [

        'Accuracy',
        'F1',
        'Precision',
        'Recall',
        'MCC'
    ],

    'Value': [

        results['eval_accuracy'],

        results['eval_f1'],

        results['eval_precision'],

        results['eval_recall'],

        results['eval_mcc']
    ]
})

print("\n================================================")
print("FINAL SUMMARY")
print("================================================")

print(summary_df)

summary_df.to_csv(

    f'{RESULT_DIR}final_summary.csv',

    index=False
)

print("\n✅ XLM-R experiment complete")
```

```python
# ============================================================
# INDICBERT EXPERIMENT — FULL PIPELINE
# ============================================================
# FEATURES:
# - Uses ai4bharat/IndicBERTv2-MLM-only
# - Optimized for Indic languages
# - Bengali-friendly
# - Stable settings
# - Full publication-ready pipeline
# ============================================================

# ============================================================
# CELL 1 — IMPORTS
# ============================================================

import os
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"

import gc
import torch
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from datasets import Dataset

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    matthews_corrcoef,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback
)

import torch.nn.functional as F

import transformers.modeling_utils as mu
mu.check_torch_load_is_safe = lambda: None

# ============================================================
# SETTINGS
# ============================================================

SEED = 42

MODEL_NAME = 'ai4bharat/IndicBERTv2-MLM-only'

DEVICE = torch.device(
    'cuda:0' if torch.cuda.is_available() else 'cpu'
)

RESULT_DIR = './BanglishDetect/results/indicbert/'

MODEL_SAVE_DIR = './best_indicbert_model'

os.makedirs(RESULT_DIR, exist_ok=True)

print("DEVICE:", DEVICE)

# ============================================================
# SEED
# ============================================================

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(SEED)

# ============================================================
# CELL 2 — LOAD DATA
# ============================================================

train_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_train.csv'
)

val_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_val.csv'
)

test_df = pd.read_csv(
    './BanglishDetect/data/processed/gold_test.csv'
)

# ============================================================
# CLEAN LABELS
# ============================================================

train_df['label'] = train_df['label'].astype(int)
val_df['label'] = val_df['label'].astype(int)
test_df['label'] = test_df['label'].astype(int)

train_df = train_df[
    train_df['label'].isin([0, 1])
]

val_df = val_df[
    val_df['label'].isin([0, 1])
]

test_df = test_df[
    test_df['label'].isin([0, 1])
]

print("\nTrain:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))

print("\nTrain labels:")
print(train_df['label'].value_counts())

# ============================================================
# CELL 3 — TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

# ============================================================
# HF DATASETS
# ============================================================

train_ds = Dataset.from_pandas(train_df)

val_ds = Dataset.from_pandas(val_df)

test_ds = Dataset.from_pandas(test_df)

# ============================================================
# TOKENIZATION
# ============================================================

def tokenize(batch):

    return tokenizer(

        batch['text'],

        truncation=True,

        padding='max_length',

        max_length=128
    )

train_ds = train_ds.map(
    tokenize,
    batched=True
)

val_ds = val_ds.map(
    tokenize,
    batched=True
)

test_ds = test_ds.map(
    tokenize,
    batched=True
)

# ============================================================
# REMOVE EXTRA COLUMNS
# ============================================================

keep_cols = [
    'input_ids',
    'attention_mask',
    'label'
]

train_ds = train_ds.remove_columns(

    [c for c in train_ds.column_names
     if c not in keep_cols]
)

val_ds = val_ds.remove_columns(

    [c for c in val_ds.column_names
     if c not in keep_cols]
)

test_ds = test_ds.remove_columns(

    [c for c in test_ds.column_names
     if c not in keep_cols]
)

# ============================================================
# TORCH FORMAT
# ============================================================

train_ds.set_format(
    type='torch',
    columns=keep_cols
)

val_ds.set_format(
    type='torch',
    columns=keep_cols
)

test_ds.set_format(
    type='torch',
    columns=keep_cols
)

print("\n✅ Tokenization complete")

# ============================================================
# CELL 4 — LOAD MODEL
# ============================================================

model = AutoModelForSequenceClassification.from_pretrained(

    MODEL_NAME,

    num_labels=2
)

model.to(DEVICE)

print("\n✅ IndicBERT loaded")

# ============================================================
# CELL 5 — METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits, labels = eval_pred

    preds = np.argmax(logits, axis=1)

    return {

        'accuracy':
            accuracy_score(labels, preds),

        'f1':
            f1_score(labels, preds),

        'precision':
            precision_score(labels, preds),

        'recall':
            recall_score(labels, preds),

        'mcc':
            matthews_corrcoef(labels, preds)
    }

# ============================================================
# CELL 6 — TRAINING ARGUMENTS
# ============================================================

training_args = TrainingArguments(

    output_dir='./indicbert_results',

    eval_strategy='epoch',

    save_strategy='epoch',

    logging_strategy='epoch',

    learning_rate=2e-5,

    per_device_train_batch_size=4,

    per_device_eval_batch_size=4,

    gradient_accumulation_steps=4,

    num_train_epochs=5,

    warmup_steps=100,

    weight_decay=0.01,

    fp16=False,

    dataloader_num_workers=0,

    load_best_model_at_end=True,

    metric_for_best_model='f1',

    greater_is_better=True,

    save_total_limit=1,

    seed=SEED,

    report_to='none'
)

print("\n✅ Training arguments ready")

# ============================================================
# CELL 7 — TRAINER
# ============================================================

trainer = Trainer(

    model=model,

    args=training_args,

    train_dataset=train_ds,

    eval_dataset=val_ds,

    compute_metrics=compute_metrics,

    callbacks=[

        EarlyStoppingCallback(
            early_stopping_patience=2
        )
    ]
)

print("\n✅ Trainer ready")

# ============================================================
# CELL 8 — TRAIN
# ============================================================

gc.collect()

torch.cuda.empty_cache()

print("\n================================================")
print("TRAINING INDICBERT")
print("================================================")

trainer.train()

print("\n✅ Training complete")

# ============================================================
# CELL 9 — EVALUATION
# ============================================================

print("\n================================================")
print("TEST EVALUATION")
print("================================================")

results = trainer.evaluate(test_ds)

for k, v in results.items():

    print(f"{k}: {v}")

# ============================================================
# SAVE RESULTS
# ============================================================

results_df = pd.DataFrame([results])

results_df.to_csv(

    f'{RESULT_DIR}indicbert_test_results.csv',

    index=False
)

print("\n✅ Results saved")

# ============================================================
# CELL 10 — SAVE MODEL
# ============================================================

trainer.save_model(MODEL_SAVE_DIR)

tokenizer.save_pretrained(MODEL_SAVE_DIR)

print("\n✅ Model saved")

# ============================================================
# CELL 11 — PREDICTIONS
# ============================================================

pred_output = trainer.predict(test_ds)

logits = pred_output.predictions

y_true = pred_output.label_ids

y_pred = np.argmax(logits, axis=1)

probs = F.softmax(
    torch.tensor(logits),
    dim=1
).numpy()

confidences = probs.max(axis=1)

# ============================================================
# CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_true,

    y_pred,

    digits=4
)

print("\n================================================")
print("CLASSIFICATION REPORT")
print("================================================")

print(report)

with open(

    f'{RESULT_DIR}classification_report.txt',

    'w',

    encoding='utf-8'

) as f:

    f.write(report)

# ============================================================
# CELL 12 — CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

fig, ax = plt.subplots(figsize=(5,5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot(ax=ax)

plt.title('IndicBERT Confusion Matrix')

plt.tight_layout()

plt.savefig(

    f'{RESULT_DIR}confusion_matrix.png',

    dpi=300,

    bbox_inches='tight'
)

plt.show()

# ============================================================
# CELL 13 — CONFIDENCE HISTOGRAM
# ============================================================

plt.figure(figsize=(8,5))

plt.hist(
    confidences,
    bins=20
)

plt.xlabel('Confidence')

plt.ylabel('Frequency')

plt.title('IndicBERT Confidence Distribution')

plt.tight_layout()

plt.savefig(

    f'{RESULT_DIR}confidence_distribution.png',

    dpi=300,

    bbox_inches='tight'
)

plt.show()

# ============================================================
# CELL 14 — ERROR ANALYSIS
# ============================================================

analysis_df = pd.DataFrame({

    'text':
        test_df['text'],

    'true_label':
        y_true,

    'prediction':
        y_pred,

    'confidence':
        confidences
})

analysis_df['correct'] = (

    analysis_df['true_label']
    ==
    analysis_df['prediction']
)

analysis_df.to_csv(

    f'{RESULT_DIR}full_error_analysis.csv',

    index=False
)

print("\n✅ Full error analysis saved")

# ============================================================
# WRONG PREDICTIONS
# ============================================================

wrong_df = analysis_df[
    analysis_df['correct'] == False
]

wrong_df.to_csv(

    f'{RESULT_DIR}wrong_predictions.csv',

    index=False
)

print("✅ Wrong predictions saved")

# ============================================================
# SAMPLE ERRORS
# ============================================================

print("\n================================================")
print("SAMPLE ERRORS")
print("================================================")

for _, row in wrong_df.head(10).iterrows():

    print("\n--------------------------------------")

    print("TEXT:")
    print(row['text'])

    print("\nTRUE:", row['true_label'])

    print("PRED:", row['prediction'])

    print("CONF:", round(row['confidence'], 4))

# ============================================================
# CELL 15 — FINAL SUMMARY
# ============================================================

summary_df = pd.DataFrame({

    'Metric': [

        'Accuracy',
        'F1',
        'Precision',
        'Recall',
        'MCC'
    ],

    'Value': [

        results['eval_accuracy'],

        results['eval_f1'],

        results['eval_precision'],

        results['eval_recall'],

        results['eval_mcc']
    ]
})

print("\n================================================")
print("FINAL SUMMARY")
print("================================================")

print(summary_df)

summary_df.to_csv(

    f'{RESULT_DIR}final_summary.csv',

    index=False
)

print("\n✅ IndicBERT experiment complete")
```
